```python
import pandas as pd

# Create a DataFrame with sample data
data = {
    'Owner': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Address': ['123 Maple St', '456 Oak St', '789 Pine St', '101 Birch St', '202 Cedar St']
}

df = pd.DataFrame(data)
df
```

```python
csv_string = df.to_csv(index=False)
```

```python
with open("OwnerAddress.csv", mode='w', encoding='utf-8') as f:
    f.write(csv_string)
```

```python
output_csv = {"url": "OwnerAddress.csv"}
```

Output Parameter
- Direction: Output
- Variable name: output_csv
- Display name: Download CSV
- Data type: File
- Default value: {"url":"OwnerAddress.csv"}
- Description: Optional

```python
# Inserted snippet writes output parameter file for web tool execution
import os
import re
import shutil
# Only run this snippet during web tool execution when env variable "ENB_JOBID" will be present
if "ENB_JOBID" in os.environ:
    out_file = output_csv["url"]
    if not re.match(r"^https://", out_file):
        out_param_dir = os.path.join(os.environ["ENB_JOBID"], "outputs")
        os.makedirs(out_param_dir, exist_ok=True)
        shutil.copy(src=out_file, dst=out_param_dir)
    out_param_name = "output_csv"
    out_param_file = os.path.join(
        os.environ["ENB_JOBID"], "value_" + out_param_name + ".dat")
    with open(out_param_file, "w") as writer:
        writer.write(str(output_csv))
```
